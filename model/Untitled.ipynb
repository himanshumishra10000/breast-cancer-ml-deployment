```python
# Import required libraries
import pandas as pd
import numpy as np

from sklearn.datasets import load_breast_cancer

# Load the dataset
dataset = load_breast_cancer()

# Create the feature dataframe
X = pd.DataFrame(
    dataset.data,
    columns=dataset.feature_names
)

# Convert target: 1 = Malignant, 0 = Benign
y = pd.Series(
    (dataset.target == 0).astype(int),
    name="target"
)

# Display dataset information
print("Dataset Shape:", X.shape)
print("\nNumber of Features:", X.shape[1])
print("\nNumber of Records:", X.shape[0])

print("\nTarget Distribution:")
print(y.value_counts())

print("\nFirst Five Records:")
display(X.head())
```

```python
# Combine input features and target
df = X.copy()
df["target"] = y

# Check missing values
print("Missing values:", df.isnull().sum().sum())

# Check duplicate records
print("Duplicate records:", df.duplicated().sum())

# Check class distribution
print("\nClass Distribution:")
print(df["target"].value_counts())

# Check data types
print("\nData Types:")
print(df.dtypes.value_counts())
```

```python
# Import train_test_split
from sklearn.model_selection import train_test_split

# Separate input features and target variable
X = df.drop("target", axis=1)
y = df["target"]

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display dataset dimensions
print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("\nTraining Target:", y_train.shape)
print("Testing Target:", y_test.shape)

# Check class distribution in training data
print("\nTraining Class Distribution:")
print(y_train.value_counts())

# Check class distribution in testing data
print("\nTesting Class Distribution:")
print(y_test.value_counts())
```

```python
# Import required libraries
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# Create Logistic Regression Pipeline
model_lr = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])

# Train the model
model_lr.fit(X_train, y_train)

# Predict target values
y_pred_lr = model_lr.predict(X_test)

# Calculate prediction probabilities
y_prob_lr = model_lr.predict_proba(X_test)[:, 1]

# Display predictions
print("Actual Values:")
print(y_test.values[:10])

print("\nPredicted Values:")
print(y_pred_lr[:10])

print("\nPredicted Probabilities:")
print(y_prob_lr[:10])
```

```python
# Import evaluation metrics
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef
)

# Calculate evaluation metrics
lr_accuracy = accuracy_score(y_test, y_pred_lr)

lr_auc = roc_auc_score(y_test, y_prob_lr)

lr_precision = precision_score(y_test, y_pred_lr)

lr_recall = recall_score(y_test, y_pred_lr)

lr_f1 = f1_score(y_test, y_pred_lr)

lr_mcc = matthews_corrcoef(y_test, y_pred_lr)

# Display results
print("LOGISTIC REGRESSION - EVALUATION RESULTS")
print("----------------------------------------")

print("Accuracy :", round(lr_accuracy, 4))
print("AUC      :", round(lr_auc, 4))
print("Precision:", round(lr_precision, 4))
print("Recall   :", round(lr_recall, 4))
print("F1 Score :", round(lr_f1, 4))
print("MCC      :", round(lr_mcc, 4))
```

```python
# Import Decision Tree Classifier
from sklearn.tree import DecisionTreeClassifier

# Create Decision Tree model
model_dt = DecisionTreeClassifier(
    random_state=42
)

# Train the model
model_dt.fit(X_train, y_train)

# Predict target values
y_pred_dt = model_dt.predict(X_test)

# Calculate prediction probabilities
y_prob_dt = model_dt.predict_proba(X_test)[:, 1]

# Display predictions
print("Actual Values:")
print(y_test.values[:10])

print("\nPredicted Values:")
print(y_pred_dt[:10])

print("\nPredicted Probabilities:")
print(y_prob_dt[:10])
```

```python
# Calculate evaluation metrics
dt_accuracy = accuracy_score(y_test, y_pred_dt)

dt_auc = roc_auc_score(y_test, y_prob_dt)

dt_precision = precision_score(y_test, y_pred_dt)

dt_recall = recall_score(y_test, y_pred_dt)

dt_f1 = f1_score(y_test, y_pred_dt)

dt_mcc = matthews_corrcoef(y_test, y_pred_dt)

# Display results
print("DECISION TREE - EVALUATION RESULTS")
print("----------------------------------")

print("Accuracy :", round(dt_accuracy, 4))
print("AUC      :", round(dt_auc, 4))
print("Precision:", round(dt_precision, 4))
print("Recall   :", round(dt_recall, 4))
print("F1 Score :", round(dt_f1, 4))
print("MCC      :", round(dt_mcc, 4))
```

```python
# Import required libraries
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# Create KNN Pipeline
model_knn = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", KNeighborsClassifier(
        n_neighbors=5
    ))
])

# Train the model
model_knn.fit(X_train, y_train)

# Predict target values
y_pred_knn = model_knn.predict(X_test)

# Calculate prediction probabilities
y_prob_knn = model_knn.predict_proba(X_test)[:, 1]

# Display predictions
print("Actual Values:")
print(y_test.values[:10])

print("\nPredicted Values:")
print(y_pred_knn[:10])

print("\nPredicted Probabilities:")
print(y_prob_knn[:10])
```

```python
# Calculate evaluation metrics
knn_accuracy = accuracy_score(y_test, y_pred_knn)

knn_auc = roc_auc_score(y_test, y_prob_knn)

knn_precision = precision_score(y_test, y_pred_knn)

knn_recall = recall_score(y_test, y_pred_knn)

knn_f1 = f1_score(y_test, y_pred_knn)

knn_mcc = matthews_corrcoef(y_test, y_pred_knn)

# Display results
print("KNN - EVALUATION RESULTS")
print("------------------------")

print("Accuracy :", round(knn_accuracy, 4))
print("AUC      :", round(knn_auc, 4))
print("Precision:", round(knn_precision, 4))
print("Recall   :", round(knn_recall, 4))
print("F1 Score :", round(knn_f1, 4))
print("MCC      :", round(knn_mcc, 4))
```

```python
# Import Gaussian Naive Bayes
from sklearn.naive_bayes import GaussianNB

# Create Gaussian Naive Bayes model
model_nb = GaussianNB()

# Train the model
model_nb.fit(X_train, y_train)

# Predict target values
y_pred_nb = model_nb.predict(X_test)

# Calculate prediction probabilities
y_prob_nb = model_nb.predict_proba(X_test)[:, 1]

# Display predictions
print("Actual Values:")
print(y_test.values[:10])

print("\nPredicted Values:")
print(y_pred_nb[:10])

print("\nPredicted Probabilities:")
print(y_prob_nb[:10])
```

```python
# Import required libraries
from sklearn.svm import SVC
from sklearn.calibration import CalibratedClassifierCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# Create SVM Pipeline
model_svm = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", CalibratedClassifierCV(
        estimator=SVC(
            kernel="rbf",
            C=1.0,
            random_state=42
        ),
        ensemble=False
    ))
])

# Train the model
model_svm.fit(X_train, y_train)

# Predict target values
y_pred_svm = model_svm.predict(X_test)

# Calculate prediction probabilities
y_prob_svm = model_svm.predict_proba(X_test)[:, 1]

# Display predictions
print("Actual Values:")
print(y_test.values[:10])

print("\nPredicted Values:")
print(y_pred_svm[:10])

print("\nPredicted Probabilities:")
print(y_prob_svm[:10])
```

```python
# Calculate evaluation metrics
svm_accuracy = accuracy_score(y_test, y_pred_svm)

svm_auc = roc_auc_score(y_test, y_prob_svm)

svm_precision = precision_score(y_test, y_pred_svm)

svm_recall = recall_score(y_test, y_pred_svm)

svm_f1 = f1_score(y_test, y_pred_svm)

svm_mcc = matthews_corrcoef(y_test, y_pred_svm)

# Display results
print("SVM - EVALUATION RESULTS")
print("-------------------------")

print("Accuracy :", round(svm_accuracy, 4))
print("AUC      :", round(svm_auc, 4))
print("Precision:", round(svm_precision, 4))
print("Recall   :", round(svm_recall, 4))
print("F1 Score :", round(svm_f1, 4))
print("MCC      :", round(svm_mcc, 4))
```
